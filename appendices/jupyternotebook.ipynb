# Adding Jupyter Notebooks Chapters

Quarto can include Jupyter notebooks (`.ipynb`) directly as book chapters. This is particularly useful when teaching materials combine narrative text, 
executable code, and visualisations created in Jupyter Notebook and you want to share it with students.

::: {.callout-note}

This chapter is actually compiled from a .ipynb file rather than a .qmd file.

:::

## Adding a Notebook to the Quarto Book

List the notebook in the `chapters` section of `_quarto.yml` just as you would a `.qmd` file (see @sec-folder-yaml).

Quarto will automatically render the notebook as a chapter in the book and include it in the table of contents and navigation.


## Controlling Code Display

Notebook cells can be shown, hidden, or folded using cell metadata.

For example, to hide a code cell while displaying its output:

```json
{
  "tags": ["hide-input"]
}
```

Common tags include:

- `hide-input` – hide code, show output
- `hide-output` – show code, hide output
- `hide-cell` – hide both code and output

## Using Notebooks with Quarto Features

Jupyter notebooks support many Quarto features, including:

- Citations
- Equations
- Figure captions
- Callouts
- Cross-references
